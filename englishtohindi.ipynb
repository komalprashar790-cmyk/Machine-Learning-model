```python
from sklearn.model_selection import train_test_split
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences


# Keep only the English and Hindi columns
data = data[['english', 'hindi']].copy()

# Remove missing values
data = data.dropna(subset=['english', 'hindi'])

# Remove extra spaces
data['english'] = data['english'].astype(str).str.strip()
data['hindi'] = data['hindi'].astype(str).str.strip()

# Remove empty sentences
data = data[(data['english'] != '') & (data['hindi'] != '')]

print("Dataset shape:", data.shape)
data.head()
```

Create training and testing set

```python
train_data, test_data = train_test_split(data,test_size = 0.2,random_state=30)
print("training_dataset",len(train_data))
print("testing_dataset",len(test_data))
```

 Tokenize English

```python
english_tokenizer = Tokenizer(
    filters='',
    oov_token='<OOV>'
)

english_tokenizer.fit_on_texts(train_data['english'])

X_train = english_tokenizer.texts_to_sequences(
    train_data['english']
)

X_test = english_tokenizer.texts_to_sequences(
    test_data['english']
)

print("English vocabulary size:",
      len(english_tokenizer.word_index) + 1)
```

Tokenize HIndi

```python
hindi_tokenizer = Tokenizer(
    filters='',
    oov_token='<OOV>'
)

hindi_tokenizer.fit_on_texts(train_data['hindi'])

y_train = hindi_tokenizer.texts_to_sequences(
    train_data['hindi']
)

y_test = hindi_tokenizer.texts_to_sequences(
    test_data['hindi']
)

print("Hindi vocabulary size:",
      len(hindi_tokenizer.word_index) + 1)
```
