```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
data = pd.read_csv("/content/Customer.csv")
print(data.head())
# Convert 'TotalCharges' to numeric, coercing errors to NaN, then fill NaN with 0
data['TotalCharges'] = pd.to_numeric(data['TotalCharges'], errors='coerce').fillna(0)
data["TotalPrice"] = data["TotalCharges"].cumsum()
print("cumsum")
print(data.head())
data.plot(x="MonthlyCharges",y="TotalCharges")
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
data = pd.read_csv("/content/world-happiness-report-2019.csv")
print(data.head())
# Fix: Fill NaN values in 'Positive affect' column with its mean
data["Ladder"] = data["Ladder"].fillna( data["Ladder"]).dropna()
data["Positive affect"] = data["Positive affect"].fillna( data["Positive affect"]).dropna()
print("after cleaning")
print(data.head())
plt.figure(figsize=(10,8))
data.plot(kind="scatter",
          x="Positive affect",
          y="Ladder",label= "positive affect vs Ladder")

data.plot(kind="scatter",x= "Freedom",y="Negative affect",label="How freedom impact Happiness")
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd

df = pd.read_csv('world-happiness-report-2019.csv')
df = df.dropna()

plt.figure(figsize=(8, 5))

plt.scatter(df['Log of GDP\nper capita'], df['Ladder'],
            label='GDP vs Happiness', color='steelblue', alpha=0.6)

plt.scatter(df['Freedom'], df['Ladder'],
            label='Freedom vs Happiness', color='coral', alpha=0.6)

plt.xlabel('Feature Value (rank)')
plt.ylabel('Happiness Rank (Ladder)')
plt.title('Feature vs Happiness')

plt.legend(loc='best', fontsize=11, title='Features', framealpha=0.7)

plt.show()
```
