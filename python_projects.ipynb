```python
# calculator
a = float(input("Take first number:"))
operators = input("Enter operator(+,-,*,/,**):" )
b = float(input("Take second number:"))

if operators == "+":
  print(a+b)
elif operators == "-":
  print(a-b)
elif operators == "*":
  print(a*b)
elif operators == "/":
  print(a/b)
elif operators == "**":
  print(a**b)

else:
  print("calculation is not possible")

```
