Bag-of-Words represents a document by counting the occurrences of each word in a vocabulary. It generally ignores word order.

```python
from sklearn.feature_extraction.text import CountVectorizer

documents = [
    "I love AI",
    "I love Python"
]

vectorizer = CountVectorizer()
X = vectorizer.fit_transform(documents)

print(vectorizer.get_feature_names_out())
print(X.toarray())
```

Word embeddings represent words as dense numerical vectors learned from text data. Words with similar meanings can have similar vector representations.

```python
import torch
import torch.nn as nn

embedding = nn.Embedding(
    num_embeddings=1000,
    embedding_dim=100
)

word_ids = torch.tensor([12, 45, 78])
vectors = embedding(word_ids)

print(vectors.shape)  # torch.Size([3, 100])
```
