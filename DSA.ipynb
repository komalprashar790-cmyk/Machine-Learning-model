````python
arr = [0,1,0,3,12]
result =[]
zero_count = 0
for x in arr:
  if x!= 0:
    result.append(x)
  elif x == 0:
    zero_count = zero_count + 1

for i in range(zero_count):
  result.append(0)
print(result)

````

Ab ek very important placement problem karte hain.

Given:

arr = [2, 7, 11, 15]
target = 9

Find two numbers whose sum is equal to target.

````python
arr =[2,7,11,15]
target = 9
for i in arr:
  for j in arr:
    if i + j == target:
      print(i,j)
````

````python
arr = [4, 6, 1, 9, 3]
target = 10
for i in arr:
  for j in arr:
    if i + j == target:
      print(i,j)
````

````python
arr = [2,7,11,15]
target = 9
for x in range(len(arr)):
  for j in range(x+1,len(arr)):
    if arr[x] + arr[j] == target:
      print(arr[x],arr[j])
````

last problem which we solve sum of two equal to specific number has time complexity = o(n^2), space complexity: o(1)



now just optimized this:
New idea: "Complement"
Agar:
target = 9
current number = 2
Toh hume kitna number chahiye?
9 - 2 = 7
Is 7 ko complement kehte hain.
Formula:
complement = target - current

2 → need 7
11 → need -2
7 → need 2
15 → need -6
Dictionary kyun?
Hum jo numbers already dekh chuke hain, unko dictionary mein store karenge.

Example:

current = 2
need = 7

7 dictionary mein nahi hai → 2 store karo.

Next:

current = 11
need = -2

-2 nahi hai → 11 store karo.

Next:

current = 7
need = 2

already dictionary mein hai!
So pair:
2 + 7 = 9

````python
arr = [2,7,11,15]
a = {} # dic
target = 9
for x in a:
  complement = target - x
  if complement in a:
    print(complement,x)
  else:
    a[x] = True
````
